```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.externals import joblib

# Load the dataset
music_data = pd.read_csv('/content/drive/MyDrive/Github/music.csv')
music_data
```

```python
X = music_data.drop(columns=['genre'])
X
```

```python
y = music_data['genre']
y
```

```python
model = DecisionTreeClassifier()
model.fit(X,y)
music_data
```

```python
# The tree helped us understand that younger men from 20 - 25 is hip hop
# Males between 26 - 30 is jazz
# females 20 - 25 is Dance
# females 26 - 30 is Acoustic
predictions = model.predict([ [21, 1], [22,0] ])
predictions
```

```python
# Calculating accuracy
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2)
model.fit(X_train, y_train)
predictions = model.predict(X_test)

joblib.dump(model, 'music-recommender.hoblib') # job live file, so we dont have to train the model every time
score = accuracy_score(y_test, predictions)
score
#note with every run the accuracy will drop from 1 to 0.75 to even 0.5 due to the size of our data
```

```python
# Visualizing a Decision Tree

import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree

music_data = pd.read_csv('/content/drive/MyDrive/Github/music.csv')
X = music_data.drop(columns=['genre'])
y = music_data['genre']

model = DecisionTreeClassifier()
model.fit(X,y)

tree.export_graphviz(model, out_file='music-recommendor.dot',
                    feature_names=['age', 'gender'],
                    class_names=sorted(y.unique()),
                    label='all',
                    rounded=True,
                    filled=True)
```
